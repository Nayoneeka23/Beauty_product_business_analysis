```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("Dataset/product_info.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.describe()
```

```python
columns = [
    "product_name",
    "brand_name",
    "primary_category",
    "secondary_category",
    "price_usd",
    "rating",
    "reviews",
    "loves_count",
    "sephora_exclusive",
    "new",
    "online_only"
]

data = df[columns].copy()

data.head()
```

```python
print("Duplicate rows:", data.duplicated().sum())
```

```python
data = data.drop_duplicates()

data = data.dropna(
    subset=["price_usd", "rating", "reviews"]
)

print("Cleaned dataset shape:", data.shape)
```

```python
print("Number of products:", len(data))
print("Number of brands:", data["brand_name"].nunique())
print("Number of categories:", data["primary_category"].nunique())

print("Average rating:", round(data["rating"].mean(), 2))
print("Median price: $", round(data["price_usd"].median(), 2))
print("Total customer loves:", data["loves_count"].sum())
```

```python
category_summary = (
    data.groupby("primary_category")
    .agg(
        products=("product_name", "count"),
        average_rating=("rating", "mean"),
        median_price=("price_usd", "median"),
        total_loves=("loves_count", "sum"),
        total_reviews=("reviews", "sum")
    )
    .reset_index()
)

category_summary = category_summary.sort_values(
    "products",
    ascending=False
)

category_summary
```

```python
fig = px.bar(
    category_summary,
    x="primary_category",
    y="products",
    title="Number of Products by Beauty Category",
    labels={
        "primary_category": "Category",
        "products": "Number of Products"
    }
)

fig.show()
```

```python
correlation = data["price_usd"].corr(data["rating"])

print(
    "Correlation between price and rating:",
    round(correlation, 3)
)
```

```python
price_plot = data[
    data["price_usd"] <= data["price_usd"].quantile(0.99)
]

fig = px.scatter(
    price_plot,
    x="price_usd",
    y="rating",
    color="primary_category",
    size="reviews",
    hover_data=[
        "product_name",
        "brand_name"
    ],
    title="Price vs Customer Rating",
    labels={
        "price_usd": "Price (USD)",
        "rating": "Customer Rating",
        "primary_category": "Category"
    },
    opacity=0.6
)

fig.show()
```

```python
brand_summary = (
    data.groupby("brand_name")
    .agg(
        products=("product_name", "count"),
        average_rating=("rating", "mean"),
        total_reviews=("reviews", "sum"),
        total_loves=("loves_count", "sum"),
        median_price=("price_usd", "median")
    )
    .reset_index()
)

established_brands = brand_summary[
    (brand_summary["products"] >= 10) &
    (brand_summary["total_reviews"] >= 500)
]

top_brands = (
    established_brands
    .sort_values("average_rating", ascending=False)
    .head(10)
)

top_brands
```

```python
fig = px.bar(
    top_brands.sort_values("average_rating"),
    x="average_rating",
    y="brand_name",
    orientation="h",
    title="Highest-Rated Established Beauty Brands",
    labels={
        "average_rating": "Average Rating",
        "brand_name": "Brand"
    }
)

fig.show()
```

```python
popularity_corr = data["loves_count"].corr(data["rating"])

print(
    "Correlation between customer loves and rating:",
    round(popularity_corr, 3)
)
```

```python
fig = px.scatter(
    data,
    x="rating",
    y="loves_count",
    color="primary_category",
    hover_data=["product_name", "brand_name"],
    log_y=True,
    title="Customer Rating vs Product Popularity",
    labels={
        "rating": "Customer Rating",
        "loves_count": "Customer Loves"
    },
    opacity=0.6
)

fig.show()
```

```python
exclusive_summary = (
    data.groupby("sephora_exclusive")
    .agg(
        products=("product_name", "count"),
        average_rating=("rating", "mean"),
        median_price=("price_usd", "median"),
        average_loves=("loves_count", "mean"),
        average_reviews=("reviews", "mean")
    )
    .reset_index()
)

exclusive_summary["product_type"] = exclusive_summary[
    "sephora_exclusive"
].map({
    0: "Non-Exclusive",
    1: "Sephora Exclusive"
})

exclusive_summary
```

```python
fig = px.bar(
    exclusive_summary,
    x="product_type",
    y="average_loves",
    title="Average Customer Interest: Exclusive vs Non-Exclusive Products",
    labels={
        "product_type": "",
        "average_loves": "Average Loves"
    }
)

fig.show()
```
