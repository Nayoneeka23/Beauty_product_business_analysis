```python
columns = [
    "product_name",
    "brand_name",
    "primary_category",
    "secondary_category",
    "price_usd",
    "rating",
    "reviews",
    "loves_count",
    "sephora_exclusive",
    "new",
    "online_only"
]

data = df[columns].copy()

data.head()
```

```python
print("Duplicate rows:", data.duplicated().sum())
```

```python
data = data.drop_duplicates()

data = data.dropna(
    subset=["price_usd", "rating", "reviews"]
)

print("Cleaned dataset shape:", data.shape)
```
